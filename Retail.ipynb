## Data Analysis & Visualization

Q1.How are Sales Changing Over Time?

```python
monthly_sales = df.groupby(df['date'].dt.to_period('M'))['total_value'].sum()
monthly_sales.index = monthly_sales.index.astype(str)

# Visualization -- Line Chart
plt.figure(figsize=(12,6))
plt.plot(
    monthly_sales.index.to_numpy(dtype=str),
    monthly_sales.to_numpy(dtype=float),
    marker='o',
    color='blue'
)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Q2. Which Categories generate the highest sales and profit?

```python
# Profit = total_value - (cost_price * quantity)
df['profit'] = df['total_value'] - (df['cost_price'] * df['quantity'])
```

```python
df.head()
```

```python
category_performance = df.groupby('category').agg(total_sales=('total_value', 'sum'), total_profit=('profit', 'sum')).sort_values(by='total_sales', ascending=False)

# Visualization -- Bar Chart
x = np.arange(len(category_performance))
width = 0.35

plt.figure(figsize=(12,6))
plt.bar(x - width/2, category_performance['total_sales'], width, label='Total Sales', color='blue')
plt.bar(x + width/2, category_performance['total_profit'], width, label='Total Profit', color='orange')
plt.xlabel('Category')
plt.ylabel('Amount')
plt.title('Category Performance: Total Sales vs Total Profit')
plt.xticks(x, [str(category) for category in category_performance.index], rotation=90)
plt.legend()
plt.tight_layout()
plt.show()
```

Q3. Which products are the most profitable?

```python
product_performance = df.groupby('subcategory').agg(total_sales=('total_value', 'sum'), total_profit=('profit', 'sum')).sort_values(by='total_profit', ascending=False)
top_10_profitable_products = product_performance.head(10)

# Visualization -- Bar Chart
plt.figure(figsize=(12,6))
plt.bar(top_10_profitable_products.index, top_10_profitable_products['total_profit'], color='green')
plt.xlabel('Product')
plt.ylabel('Total Profit')
plt.title('Top 10 Profitable Products')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Q4. Which Stores and Cities are performing best?

```python
# Store Performance
store_performance = df.groupby('store_name')['total_value'].sum().sort_values(ascending=False)

# Visualization -- Bar Chart
plt.figure(figsize=(12,6))
plt.bar(
    store_performance.index.to_numpy(dtype=str),
    store_performance.to_numpy(dtype=float),
    color='purple'
)
plt.xlabel('Store')
plt.ylabel('Total Sales')
plt.title('Store Performance: Store Wise Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# City Performance
city_performance = df.groupby('city')['total_value'].sum().sort_values(ascending=False)

# Visualization -- Bar Chart
plt.figure(figsize=(12,6))
plt.bar(
    city_performance.index.to_numpy(dtype=str),
    city_performance.to_numpy(dtype=float),
    color='teal'
)
plt.xlabel('City')
plt.ylabel('Total Sales')   
plt.title('City Performance: City Wise Total Sales')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Q5. Which Sales Channel contributes the most revenue?

```python
# Channel Performance
channel_performance = df.groupby('channel')['total_value'].sum().sort_values(ascending=False)

# Visualization -- Bar Chart
plt.figure(figsize=(12,6))
plt.bar(
    channel_performance.index.to_numpy(dtype=str),
    channel_performance.to_numpy(dtype=float),
    color='teal'
)
plt.xlabel('Channel')
plt.ylabel('Total Sales')   
plt.title('Channel Wise Revenue Contribution')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

# Ad-Hoc Analysis

Q1. Which high sales product are at risk of stock out?

```python
product_sales = df.groupby('sku_id')['quantity'].sum().reset_index()
```

```python
risk_df = product_sales.merge(idf, on='sku_id', how='left')
```

```python
risk_df
```

```python
high_sales_low_stock = risk_df[(risk_df['quantity'] > risk_df['quantity'].median()) & (risk_df['stock_on_hand'] < risk_df['stock_on_hand'].median())]
high_sales_low_stock
```

Q2. Which products are the below their re-order point?


```python
low_stock = idf[idf['stock_on_hand'] < idf['reorder_point']]
low_stock
```

Q3. Which products have high sales but low profit?

```python
product_summary = df.groupby('sku_name').agg(total_sales=('total_value', 'sum'), total_profit=('profit', 'sum'))
product_summary
```

```python
high_sales_low_profit = product_summary.loc[(product_summary['total_sales'] > product_summary['total_sales'].median()) & (product_summary['total_profit'] < product_summary['total_profit'].median())]
high_sales_low_profit
```
